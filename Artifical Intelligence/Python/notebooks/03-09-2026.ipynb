## Slicing 

# what is slicing 
    In Python, slicing is an operation that extracts a specific portion (or subset) of a sequence—such as a list, string, or tuple—without modifying the original data. It creates a new shallow copy of the selected elements based on their index positions

### Structure :

    [start_loc:end_loc:step]  >>>>  [0:-1:1]
